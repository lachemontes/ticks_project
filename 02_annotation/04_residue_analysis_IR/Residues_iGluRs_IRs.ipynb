```python
from Bio import AlignIO
from Bio.Seq import Seq
from openpyxl import Workbook
```

```python
# Cargar el archivo de alineamiento en formato FASTA
alignment_file = "aln_soft_resudues2.fasta"
alignment = AlignIO.read(alignment_file, "fasta")
```

```python
# Definir las posiciones y los aminoácidos esperados
#positions = [537, 717, 717, 748, 748] Alineamiento 1
#positions = [1093, 1403, 1403, 1468, 1468] Alineamiento 2

#positions = [551, 742, 742, 787, 787]
#expected_aa = ["R", "T/S", "S/T", "D/E", "D/E"]

# Crear la tabla de salida
#output_table = []
```

```python
# Definir las posiciones y los aminoácidos esperados
positions = [391, 455, 455, 496, 496]
expected_aa = ["R", "T/S", "S/T", "D/E", "E/D"]

# Crear la tabla de salida
output_table = []

# Recorrer cada secuencia en el alineamiento
for record in alignment:
    sequence_id = record.id
    row = [sequence_id]

    # Recorrer cada posición y verificar el aminoácido
    for pos, aa_expected in zip(positions, expected_aa):
        if pos <= len(record.seq):
            aa = record.seq[pos - 1]

            # Verificar si el aminoácido coincide con los esperados en la posición actual
            if aa in aa_expected.split('/'):
                row.append("1 (" + aa + ")")  # Colocar "1" y el aminoácido en paréntesis si hay un match
            elif aa == "-":
                row.append("-")  # Colocar "-" si es un gap
            else:
                row.append("0 (" + aa + ")")  # Colocar "0" y el aminoácido en paréntesis si no hay match
        else:
            row.append("0")  # Colocar "0" si la posición está fuera del rango de la secuencia

    output_table.append(row)

```

```python
# Imprimir la tabla de salida
print("ID/Position\t" + "\t".join(map(str, positions)))
for row in output_table:
    print("\t".join(map(str, row)))


```

```python
# Crear un libro de Excel y una hoja de trabajo
wb = Workbook()
ws = wb.active
ws.title = "Alignment Analysis"

# Escribir encabezados
headers = ["Seq_ID", "Sequence"] + [f"Pos_{pos}" for pos in positions]
ws.append(headers)

# Escribir datos en la hoja de trabajo
for row in output_table:
    ws.append(row)

# Guardar el archivo de Excel
excel_output_file = "residues_iGluR_IRsZAIDE.xlsx"
wb.save(excel_output_file)

print(f"Output guardado en {excel_output_file}")
```
